### Étape 6 — Regrouper
**Question :** quel chiffre d'affaires chaque mois ? Il faut d'abord une colonne « mois », puis **regrouper** les commandes par mois.

**6a. La colonne mois.** `.dt.to_period("M")` garde l'année et le mois d'une date (« 2025-12 »). `.astype(str)` le transforme en texte lisible. ⚠️ **Cette ligne modifie `df_database`** : encore une colonne de plus.

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
df_database["mois"] = df_database["date_commande"].dt.to_period("M").astype(str)

print(df_database.shape)      # 12 colonnes avant, 13 maintenant
df_database.head()            # la colonne "mois" apparaît tout à droite
```

**6b. Regrouper avec `groupby`.** Se lit en 3 morceaux, de gauche à droite :

| Morceau | Ce qu'il fait | Dans Excel |
|---|---|---|
| `df_database.groupby("mois", as_index=False)` | fait un paquet de lignes par mois | les lignes d'un tableau croisé |
| `["montant"]` | dans chaque paquet, prend la colonne montant | le champ « valeurs » |
| `.sum()` | additionne dans chaque paquet | « somme de » |

**`as_index=False`** veut dire : « garde `mois` comme une **colonne normale** du résultat ». On obtient ainsi un **tableau simple à deux colonnes**, `mois` et `montant`, qu'on range dans `df_ca_par_mois`. **`df_database` n'est pas modifié** : on crée un nouveau tableau.

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
df_ca_par_mois = df_database.groupby("mois", as_index=False)["montant"].sum()

print(df_ca_par_mois.shape)       # 24 lignes (24 mois), 2 colonnes
df_ca_par_mois.round(0)           # 🔮 quel mois est le plus fort ? le plus faible ?
```

👉 Le tableau est passé de **4 475 lignes** (une par commande) à **24 lignes** (une par mois). Le pic de **décembre** (Noël), le creux d'**août** (vacances), une hausse en **mai** (fête des mères). Et d'une année sur l'autre, les ventes **augmentent**.

### 🪪 Carte d'identité
| | |

| **Né en** | 2006, pour le calcul scientifique |

| **Lien avec l'IA** | **le socle de tout le reste** : les modèles d'IA (scikit-learn, PyTorch, TensorFlow) calculent sur des arrays. Une image, un son, un texte y deviennent des listes de nombres |

### pandas ou NumPy : quelle différence ?
| | **pandas** | **NumPy** |
|---|---|---|
| L'objet principal | un **tableau** (DataFrame) : lignes et colonnes | un **array** : une liste de nombres |
| Des noms de colonnes ? | **oui** : `df_database["montant"]` | **non** : juste des valeurs, repérées par leur position |
| Des étiquettes de lignes ? | **oui** : l'index | **non** : position 0, 1, 2… |
| Mélanger texte, nombres et dates ? | **oui**, une colonne de chaque | **non**, un seul type à la fois (en pratique : des nombres) |
| Le point fort | **organiser** et **résumer** des données | **calculer** très vite |
| Préfixe dans ce notebook | `df_`, `serie_` | `np_` |

👉 **Les deux travaillent ensemble :** pandas est construit **sur** NumPy. Chaque colonne d'un DataFrame contient un array NumPy. On passe de l'un à l'autre en une ligne.

### Étape 1 — Créer un array
`np.array([ … ])` transforme une liste Python (M2.5) en array NumPy. La grande différence : **une opération s'applique à tous les éléments d'un coup**.

```python
# 🧪 EXEMPLE — données inventées pour comprendre, sans lien avec database.csv
liste_prix = [24.90, 18.00, 45.00]          # une liste Python
np_prix = np.array([24.90, 18.00, 45.00])    # un array NumPy

print(liste_prix * 2)       # 🔮 avec une liste ?
print(np_prix * 2)          # 🔮 avec un array ?
```

👉 Avec une **liste**, `* 2` **répète** la liste (6 éléments). Avec un **array**, `* 2` **multiplie chaque prix** par 2. C'est tout l'intérêt de NumPy.

### Étape 2 — Passer de pandas à NumPy

### Étape 4 — Pourquoi NumPy est rapide : la vectorisation
**Question :** appliquer une remise de 15 % à **1 million** de montants. Deux façons :
- **une boucle `for`** (M2.8) : Python traite les montants **un par un** ;
- **NumPy** : une seule opération sur tout l'array. On appelle ça **vectoriser**.

`np.resize(np_montants, 1_000_000)` répète nos 4 475 montants jusqu'à en avoir 1 million. `time.perf_counter()` lit un chronomètre : on le lit avant et après, la différence donne la durée.

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
import time

np_un_million = np.resize(np_montants, 1_000_000)

# 1. Avec une boucle for
debut = time.perf_counter()
liste_remises = []
for montant in np_un_million:
    liste_remises.append(montant * 0.85)
duree_boucle = time.perf_counter() - debut

# 2. Avec NumPy
debut = time.perf_counter()
np_remises = np_un_million * 0.85
duree_numpy = time.perf_counter() - debut

print("Boucle :", round(duree_boucle, 3), "seconde")
print("NumPy  :", round(duree_numpy, 4), "seconde")
print("NumPy est", round(duree_boucle / duree_numpy), "fois plus rapide")
```

👉 Même résultat, mais NumPy est **des dizaines de fois plus rapide** (environ 50 fois sur Colab ; le chiffre varie d'une machine à l'autre). Sur les milliards de calculs d'un modèle d'IA, c'est la différence entre une heure et un mois.

### Étape 0 — Charger le fichier

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
# 🔁 ÉTAPE 0 — charger le fichier. Elle rend ce module indépendant des autres : exécute-la toujours en premier.

import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error

URL_DATA = "https://raw.githubusercontent.com/SAMIMZOUGHI/MBADMB_python/main/data/"
df_database = pd.read_csv(URL_DATA + "database.csv")

print("✅ df_database chargé :", df_database.shape[0], "lignes,", df_database.shape[1], "colonnes")
```

### Étape 1 — Ce qu'on sait (X) et ce qu'on veut prédire (y)
- **`X`** : les **variables** qui servent à prédire. Pour commencer, une seule : `nb_articles`. ⚠️ **Deux paires de crochets** `[["nb_articles"]]` : scikit-learn veut un **tableau** (`df_`), même avec une seule colonne.
- **`y`** : la **cible**, ce qu'on veut prédire : `montant`. **Une paire de crochets** : une seule colonne (`serie_`).

Rien n'est modifié dans `df_database` : on en extrait deux morceaux.

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
df_X = df_database[["nb_articles"]]       # [[ ]] : un tableau d'une colonne
serie_y = df_database["montant"]          # [ ]   : une seule colonne

print("X :", df_X.shape, "| y :", serie_y.shape)
print(df_X.head(3))
print(serie_y.head(3))
```

### Étape 2 — Mettre des commandes de côté pour le test
Un modèle qu'on évalue sur les exemples qu'il a appris, c'est un élève qu'on interroge sur les questions qu'il a déjà vues : il aura toujours bonne note. On coupe donc le fichier en deux :
- **80 % pour apprendre** (`train`) ;
- **20 % pour tester** (`test`) : des commandes que le modèle **n'aura jamais vues**.

`random_state=42` fixe le tirage au sort : tout le monde obtient le même découpage.

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
df_X_train, df_X_test, serie_y_train, serie_y_test = train_test_split(
    df_X, serie_y, test_size=0.2, random_state=42
)

print("Pour apprendre :", df_X_train.shape[0], "commandes")
print("Pour tester    :", df_X_test.shape[0], "commandes")
```

### Étape 3 — Le modèle apprend : `.fit`
Une **régression linéaire** cherche la droite qui colle le mieux aux exemples :

**montant prédit = base + prix par article × nb_articles**

On crée le modèle (vide), puis `.fit` le fait apprendre. Le nom commence par `sklearn_` : c'est un **modèle** scikit-learn.

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
sklearn_modele = LinearRegression()                      # un modèle vide
sklearn_modele.fit(df_X_train, serie_y_train)            # il apprend sur les 80 %

print("Base :", round(sklearn_modele.intercept_, 2), "€")
print("Par article :", round(sklearn_modele.coef_[0], 2), "€")
```

👉 **Lecture business :** une commande, c'est environ **44,91 € de base**, puis **7,43 € par article** en plus. Le modèle a trouvé ces deux chiffres seul, à partir de 3 580 commandes.

📂 On prédit maintenant les **895 commandes de test**, que le modèle n'a jamais vues, et on compare à leur vrai montant. L'**erreur moyenne** (`mean_absolute_error`) dit de combien d'euros il se trompe en moyenne, dans un sens ou dans l'autre.

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
np_predictions_test = sklearn_modele.predict(df_X_test)

erreur_moyenne = mean_absolute_error(serie_y_test, np_predictions_test)
print("Erreur moyenne :", round(erreur_moyenne, 2), "€")

# 5 commandes de test : vrai montant et montant prédit, côte à côte
df_comparaison = pd.DataFrame({"vrai_montant": serie_y_test.to_numpy()[:5],
                               "montant_predit": np_predictions_test[:5].round(2)})
df_comparaison
```

### Étape 3 — Préparer une ligne par ville
`database.csv` a **4 475 lignes** (une par commande), mais il n'y a que **15 villes**. Pour la carte, il faut **une ligne par ville** avec son CA, ses coordonnées et un indicateur « a une boutique ».

**3a.** On ajoute à `df_database` une colonne `en_boutique` : 1 si la commande a été passée en boutique, 0 sinon. ⚠️ **Cette ligne modifie `df_database`** : on affiche sa taille et ses premières lignes.

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
df_database["en_boutique"] = (df_database["canal"] == "boutique").astype(int)    # True / False devient 1 / 0

print(df_database.shape)      # 11 colonnes au départ, 12 maintenant
df_database.head()            # la colonne "en_boutique" apparaît tout à droite
```

**3b.** Un `groupby` avec `as_index=False` (la ville reste une colonne) et `.agg` (plusieurs calculs d'un coup, chacun dans une colonne nommée). Modèle de chaque ligne : `nouvelle_colonne=("colonne_de_départ", "calcul")`.

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
df_villes = df_database.groupby("ville", as_index=False).agg(
    ca=("montant", "sum"),                      # le CA de la ville
    nb_commandes=("id_commande", "count"),      # le nombre de commandes
    latitude=("latitude", "first"),             # "first" : la première valeur (identique pour toute la ville)
    longitude=("longitude", "first"),
    nb_en_boutique=("en_boutique", "sum"),      # le nombre de commandes passées en boutique
)
df_villes["a_une_boutique"] = df_villes["nb_en_boutique"] > 0     # True si au moins une commande en boutique

print(df_villes.shape)        # 15 villes
df_villes.round(1)
```

👉 `df_villes` : **15 lignes**, une par ville. La dernière ligne de la cellule ajoute la colonne `a_une_boutique` (`True` / `False`) à `df_villes`. `df_database` n'est pas modifié par le `groupby`.

**Comment il pense :** il découpe une série en morceaux simples :
- une **tendance** (les ventes montent ou baissent sur le long terme) ;
- des **saisonnalités** (le cycle de l'année, celui de la semaine) ;
- du **bruit** (ce qu'il ne sait pas expliquer).

**La question :** combien Amora Zone vendra-t-elle dans les 3 prochains mois, Noël compris ?

### Étape 0 — Charger le fichier

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
# 🔁 ÉTAPE 0 — charger le fichier. Elle rend ce module indépendant des autres : exécute-la toujours en premier.

import pandas as pd

URL_DATA = "https://raw.githubusercontent.com/SAMIMZOUGHI/MBADMB_python/main/data/"
df_database = pd.read_csv(URL_DATA + "database.csv")

print("✅ df_database chargé :", df_database.shape[0], "lignes,", df_database.shape[1], "colonnes")
```

### Étape 1 — Le CA de chaque jour
Comme en M5 : une vraie date, une colonne « jour », puis un `groupby` **avec `as_index=False`** (le jour reste une colonne) qui additionne les montants de chaque jour dans une colonne `ca`. ⚠️ Les deux premières lignes **modifient `df_database`** ; le `groupby` crée un nouveau tableau, `df_ca_par_jour`.

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
df_database["date_commande"] = pd.to_datetime(df_database["date_commande"])
df_database["jour_commande"] = df_database["date_commande"].dt.date

df_ca_par_jour = df_database.groupby("jour_commande", as_index=False).agg(ca=("montant", "sum"))

print(df_ca_par_jour.shape)       # environ 720 jours, 2 colonnes
df_ca_par_jour.head()
```

### Étape 2 — Le format imposé par Prophet
Prophet exige un tableau à **deux colonnes**, nommées **exactement** :
- **`ds`** : la date (au format date de pandas) ;
- **`y`** : la valeur à prévoir.

On crée ce tableau, `df_prophet_ventes`, à partir de `df_ca_par_jour`.

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
df_prophet_ventes = pd.DataFrame({
    "ds": pd.to_datetime(df_ca_par_jour["jour_commande"]),     # la date
    "y": df_ca_par_jour["ca"],                                 # le CA du jour
})
print(len(df_prophet_ventes), "jours")
df_prophet_ventes.tail()
```

### Étape 3 — Entraîner et prévoir
| Ligne | Rôle |
|---|---|
| `prophet_modele = Prophet(yearly_seasonality=True)` | crée le modèle, en lui demandant d'apprendre le **cycle de l'année** |
| `prophet_modele.fit(df_prophet_ventes)` | **l'entraîne** : il apprend la tendance et les saisonnalités |
| `prophet_modele.make_future_dataframe(periods=90)` | prépare les dates : les 2 ans connus + 90 jours |
| `prophet_modele.predict(…)` | **prévoit** chaque jour |

⚠️ **Le piège de `yearly_seasonality=True` :** par défaut, Prophet n'apprend le cycle de l'année **que s'il a au moins 2 ans d'historique**. Ici on a 729 jours, un de moins. Sans cette option, il ignorerait Noël et prévoirait une ligne presque plate. C'est pour ça qu'il affiche un avertissement (« less than 730 days ») : on lui force la main en connaissance de cause. **Toujours lire les avertissements d'un modèle.**

L'entraînement prend quelques secondes ; des messages techniques peuvent s'afficher.

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
from prophet import Prophet

prophet_modele = Prophet(yearly_seasonality=True)
prophet_modele.fit(df_prophet_ventes)

df_prophet_futur = prophet_modele.make_future_dataframe(periods=90)     # + 90 jours
df_prophet_prevision = prophet_modele.predict(df_prophet_futur)

df_prophet_prevision[["ds", "yhat", "yhat_lower", "yhat_upper"]].tail()
```

| Colonne | Signification |
|---|---|
| `yhat` | la prévision (*ŷ*, « y chapeau ») |
| `yhat_lower` / `yhat_upper` | la fourchette : dans 80 % des cas, la vraie valeur tombera entre les deux |

### Étape 4 — Voir la prévision

```python
# 📂 DONNÉES RÉELLES — cette cellule travaille sur le fichier database.csv
import matplotlib.pyplot as plt

prophet_modele.plot(df_prophet_prevision, xlabel="Date", ylabel="CA par jour (€)")
plt.title("Prévision du CA journalier (zone bleue = fourchette)")
plt.show()

prophet_modele.plot_components(df_prophet_prevision)      # la tendance, la semaine, l'année
plt.show()
```

👉 **Les points noirs** = les ventes réelles de `database.csv`. **La ligne bleue** = ce que le modèle a appris, prolongée de 90 jours. **La zone bleu clair** = la fourchette d'incertitude.

Le second graphique montre ce que Prophet a trouvé **seul** : une **tendance** à la hausse, un **cycle annuel** (pic de fin d'année, creux d'août) et un **cycle hebdomadaire** (plus de ventes le week-end).
