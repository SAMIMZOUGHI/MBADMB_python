# M2.2 · Les types de données : Python doit savoir ce qu'il manipule
**Le problème business :** la fiche d'un client d'Amora Zone contient des choses très différentes : un prénom, un nombre de commandes, un montant en euros, un « oui / non » pour la newsletter. On **additionne** des montants, mais on n'additionne pas des prénoms. Pour savoir ce qu'il a le droit de faire, Python range chaque donnée dans un **type**.

### Les 4 types de base
| Type | Nom en Python | Exemple chez Amora Zone | Ce qu'on en fait |
|---|---|---|---|
| nombre entier | `int` | `3` commandes | compter, additionner |
| nombre à virgule | `float` | `89.70` € | calculer des montants, des moyennes |
| texte | `str` | `"Léa"`, `"Lyon"` | afficher, personnaliser un message |
| vrai / faux | `bool` | `True` (inscription à la newsletter) | décider (oui / non) |

La fonction `type( )` te dit quel est le type d'une donnée.

````python
# 🧪 EXEMPLE — données inventées pour comprendre, sans lien avec database.csv
print(type(3))
print(type(89.70))
print(type("Léa"))
print(type(True))
````

### Comment Python reconnaît le type ?
- des **guillemets** → c'est du texte (`str`), quoi qu'il y ait dedans ;
- un **point** → c'est un nombre à virgule (`float`) ;
- un nombre **sans point** → c'est un entier (`int`) ;
- `True` ou `False` **sans guillemets**, avec une majuscule → c'est un booléen (`bool`).

🔮 **Prédis le type de chaque valeur AVANT d'exécuter :**

````python
# 🧪 EXEMPLE — données inventées pour comprendre, sans lien avec database.csv
print(type(4))
print(type(4.0))
print(type("4"))
print(type("True"))
print(type(False))
print(type("89.70"))
````

👉 `"4"` et `"89.70"` sont du **texte** : les guillemets gagnent toujours. Et `"True"` entre guillemets n'est pas un booléen, c'est un mot.

### Le même symbole, des résultats différents
Le type change le comportement des opérations. 🔮 **Prédis les 3 résultats :**

````python
# 🧪 EXEMPLE — données inventées pour comprendre, sans lien avec database.csv
print(3 + 3)
print("3" + "3")
print("Ah" * 3)
````

👉 Avec des nombres, `+` **additionne**. Avec du texte, `+` **colle** les morceaux. C'est pour ça qu'un montant importé comme du texte est un piège : `"45" + "10"` donne `"4510"`, pas `55` !

### Et le code postal ?
Un code postal **ressemble** à un nombre, mais on ne l'additionne jamais, et certains commencent par 0 (`01000`, Bourg-en-Bresse). On le range donc en **texte** : `"01000"`.
🔮 Que se passe-t-il si on l'écrit comme un nombre ?

````python
# 🧪 EXEMPLE — données inventées pour comprendre, sans lien avec database.csv
code_postal = 01000
````

👉 Python refuse un nombre qui commence par 0. **Règle pratique : un identifiant (code postal, téléphone, numéro de commande) se range en texte.**

# M2.3 · Les variables et la mémoire du notebook
**Le problème business :** Léa a commandé **3 articles pour 89,70 €**, avec un **code promo de 15 %**. On va réutiliser ces chiffres plusieurs fois : plutôt que de les retaper, on leur donne un **nom**.

### Une variable, c'est quoi ?
Une **variable** est une **boîte avec une étiquette** : l'étiquette est le nom, le contenu est la valeur.

```
montant_panier = 89.70
└─ le nom ──┘  │  └ la valeur
               └ « range dans »
```

Trois règles suffisent pour tout comprendre :
1. **Le `=` veut dire « range dans »**, pas « égal » comme en maths. `montant_panier = 89.70` se lit : « range 89.70 dans la boîte `montant_panier` ».
2. **Python lit d'abord ce qui est à droite du `=`**, calcule la valeur, puis la range dans la boîte de gauche.
3. **Une boîte ne contient qu'une valeur à la fois.** Si on range une nouvelle valeur, l'ancienne est remplacée et perdue.

Une fois la boîte remplie, on écrit son **nom** partout où on aurait écrit la valeur.

````python
# 🧪 EXEMPLE — données inventées pour comprendre, sans lien avec database.csv
client = "Léa"
nb_articles = 3
montant_panier = 89.70

print(client, "a commandé", nb_articles, "articles pour", montant_panier, "€")
print(type(montant_panier))      # une variable a le type de la valeur qu'elle contient (M2.2)
````

🔮 **Prédis ce que va afficher la dernière ligne de la cellule suivante :**

````python
# 🧪 EXEMPLE — données inventées pour comprendre, sans lien avec database.csv
stock = 10
print("Avant la vente :", stock)
stock = stock - 3      # on vend 3 crèmes
print("Après la vente :", stock)
````

👉 La 2ᵉ ligne choque souvent, car en maths `stock = stock - 3` est impossible. Relis-la avec les règles 2 et 3 :
1. Python lit la droite : `stock - 3`, avec `stock` qui vaut 10 → **7** ;
2. il range 7 dans la boîte `stock` : le 10 est remplacé.

En français : « le nouveau stock, c'est l'ancien stock moins 3 ».

### Nom d'une variable ou texte ?
Les guillemets changent tout :
- `client` **sans guillemets** = le **nom d'une boîte** : Python va chercher ce qu'elle contient ;
- `"client"` **avec guillemets** = le **mot** « client », tel quel.

### 🧠 La mémoire du notebook
C'est **la** notion à comprendre pour ne pas se perdre dans un notebook.

**Toutes les cellules de code partagent la même mémoire.** Une variable créée dans une cellule reste disponible pour **toutes les cellules exécutées ensuite**, même 30 cellules plus bas. C'est pour ça que la cellule de vérification de l'exercice 2 a pu lire `prenom` et `nb_commandes`.

Une image utile : **le notebook est comme un classeur Excel où une cellule réutilise le résultat d'une autre**. Avec une grosse différence : dans Excel, tout se recalcule automatiquement ; dans un notebook, **rien ne se recalcule tant que tu n'exécutes pas la cellule**.

Ce qui compte n'est donc **pas la position** de la cellule sur la page, mais **l'ordre dans lequel tu les as exécutées** (le numéro `[n]` à gauche). Vérifions-le en 3 étapes.

# M2.7 · Les conditions : faire décider Python
**Le problème business :** la livraison est **offerte dès 50 €**, sinon elle coûte **4,90 €**. Le programme doit choisir tout seul.

```
if condition:
    ce qu'on fait si c'est vrai
else:
    ce qu'on fait sinon
```
⚠️ Deux détails qui comptent : les **deux-points** `:` à la fin de la ligne, et le **décalage** (4 espaces, ou la touche Tab) des lignes en dessous. Le décalage dit à Python ce qui « dépend » du `if`.

````python
# 🧪 EXEMPLE — données inventées pour comprendre, sans lien avec database.csv
montant = 42.00

if montant >= 50:
    frais_livraison = 0
    print("🎁 Livraison offerte !")
else:
    frais_livraison = 4.90
    print("Plus que", round(50 - montant, 2), "€ pour la livraison offerte")

print("Total à payer :", montant + frais_livraison, "€")
````

👉 La dernière ligne (`print("Total à payer…`) n'est **pas décalée** : elle ne dépend ni du `if` ni du `else`, elle s'exécute dans tous les cas.

### Plus de deux cas : `elif`
`elif` (= « sinon, si ») ajoute des cas intermédiaires. Python teste les conditions **dans l'ordre** et s'arrête à la première vraie.

````python
# 🧪 EXEMPLE — données inventées pour comprendre, sans lien avec database.csv
montant = 130

if montant >= 100:
    cadeau = "un sérum offert"
elif montant >= 50:
    cadeau = "un échantillon offert"
else:
    cadeau = "pas de cadeau"

print("Pour", montant, "€ :", cadeau)
````

# M2.8 · Les boucles : répéter une action
**Le problème business :** on sait traiter **une** commande ou **un** client. Mais Amora Zone en a des centaines. La **boucle `for`** répète la même action **pour chaque élément** d'une liste.

```
for element in liste:
    ce qu'on fait pour chaque élément
```
Mêmes règles que le `if` : deux-points et décalage. À chaque tour, la variable `element` contient l'élément suivant de la liste.

### Additionner dans une boucle : l'« accumulateur »
**Question business :** combien des 7 commandes du 25 février donnent droit à la livraison offerte (≥ 50 €) ?

Pour compter, on reproduit ce que ferait une personne avec un **compteur manuel** à l'entrée d'une boutique :
1. **avant l'ouverture**, elle remet le compteur à 0 ;
2. **pendant la journée**, elle appuie une fois chaque fois qu'un client remplit la condition ;
3. **à la fermeture**, elle lit le chiffre affiché.

En Python, le compteur est une variable qu'on appelle un **accumulateur**.
🔮 **Prédis le résultat avant d'exécuter :**

````python
# 📂 DONNÉES RÉELLES — valeurs recopiées du fichier database.csv (les vraies commandes d'Amora Zone)
nb_livraisons_offertes = 0          # 1. AVANT la boucle : on part de 0

for montant in commandes_jour:
    if montant >= 50:               # 2. DANS la boucle : la condition est testée à chaque tour
        nb_livraisons_offertes = nb_livraisons_offertes + 1   # « un clic »

print("Livraisons offertes :", nb_livraisons_offertes)        # 3. APRÈS la boucle : on lit le résultat
````

### Ce qui se passe, tour par tour
| Tour | `montant` | `montant >= 50` ? | Compteur avant | Calcul | Compteur après |
|---|---|---|---|---|---|
| 1 | 110.12 | oui | 0 | 0 + 1 | 1 |
| 2 | 116.05 | oui | 1 | 1 + 1 | 2 |
| 3 | 25.65 | non | 2 | rien | 2 |
| 4 | 46.18 | non | 2 | rien | 2 |
| 5 | 82.54 | oui | 2 | 2 + 1 | 3 |
| 6 | 32.45 | non | 3 | rien | 3 |
| 7 | 117.94 | oui | 3 | 3 + 1 | **4** |

Après le 7ᵉ tour, la liste est épuisée : la boucle s'arrête et le `print` affiche **4**.

**La ligne clé :** `nb_livraisons_offertes = nb_livraisons_offertes + 1`. C'est la règle de M2.3 : Python lit d'abord la droite (la valeur actuelle du compteur, plus 1), puis range le résultat dans la même boîte. En français : « le nouveau compteur, c'est l'ancien plus un ».

La cellule suivante affiche le compteur **à chaque tour** pour que tu le voies évoluer :

````python
# 📂 DONNÉES RÉELLES — valeurs recopiées du fichier database.csv (les vraies commandes d'Amora Zone)
nb_livraisons_offertes = 0

for montant in commandes_jour:
    if montant >= 50:
        nb_livraisons_offertes = nb_livraisons_offertes + 1
    print(f"Commande de {montant} € → compteur = {nb_livraisons_offertes}")   # décalé : affiché à chaque tour
````

### La recette en 3 temps
L'emplacement de chaque ligne (son décalage) décide **combien de fois** elle s'exécute :

| Où | Quoi | Combien de fois |
|---|---|---|
| **avant** la boucle (pas décalé) | créer l'accumulateur à 0 | 1 fois |
| **dans** la boucle (décalé) | ajouter si besoin | à chaque tour |
| **après** la boucle (pas décalé) | afficher le résultat | 1 fois |

🔴 **L'erreur la plus fréquente** : créer l'accumulateur **dans** la boucle. Il est alors remis à 0 à chaque tour, et on ne garde que le dernier. 🔮 Prédis le résultat de la cellule suivante, puis exécute-la :

````python
# 📂 DONNÉES RÉELLES — valeurs recopiées du fichier database.csv (les vraies commandes d'Amora Zone)
for montant in commandes_jour:
    nb_livraisons_offertes = 0          # ❌ remis à 0 à CHAQUE tour
    if montant >= 50:
        nb_livraisons_offertes = nb_livraisons_offertes + 1

print("Livraisons offertes :", nb_livraisons_offertes)   # affiche 1 au lieu de 4
````

👉 Au dernier tour, le compteur repart de 0, puis 117.94 ≥ 50 le fait passer à 1. Les 3 clics précédents ont été effacés. Il n'y a **pas de message d'erreur** : le résultat est simplement faux. D'où l'intérêt de **prédire** avant d'exécuter.

### Compter ou additionner : la même recette
Pour **additionner** des montants au lieu de compter, on ajoute la valeur elle-même au lieu de 1 :

````python
# 📂 DONNÉES RÉELLES — valeurs recopiées du fichier database.csv (les vraies commandes d'Amora Zone)
ca_jour = 0                              # avant : on part de 0 €

for montant in commandes_jour:
    ca_jour = ca_jour + montant       # dans : on ajoute le montant de chaque commande

print("CA du jour :", round(ca_jour, 2), "€")         # après : 530.93
````

👉 Tour par tour : 0 → 110.12 → 226.17 → 251.82 → 298.00 → 380.54 → 412.99 → **530.93**. C'est exactement ce que fait `sum(commandes_jour)` en coulisses.

💡 Si tu affiches les totaux intermédiaires, tu verras parfois `226.17000000000002` au lieu de `226.17`. Ce n'est pas une erreur de ton code : l'ordinateur stocke les nombres à virgule de façon approchée. `round( , 2)` règle l'affichage.

# M2.9 · Visualiser : du chiffre au graphique
**Le problème business :** une liste de 7 montants, ça se lit. Une liste de 12 mois de chiffre d'affaires, beaucoup moins. Un **graphique** fait apparaître en une seconde ce que les chiffres cachent : un pic, un creux, une tendance.

### Ouvrir l'appli de graphiques : `import`
Python ne sait pas dessiner tout seul. On utilise une **librairie**, **Plotly**, déjà installée dans Colab. Souviens-toi de la métaphore : la librairie est une **appli** ; `import`, c'est **l'ouvrir**.

```
import plotly.express as px
└ ouvre ┘ └─ l'appli ──┘    └ son surnom : on écrira px.… ensuite
```

Un graphique Plotly a besoin de **deux listes de même longueur** : `x` (ce qu'on met en bas, les étiquettes) et `y` (la hauteur, les valeurs).

📂 Ce module travaille sur **de vraies valeurs recopiées de `database.csv`**. Il est indépendant : sa première cellule recrée tout ce dont il a besoin.

````python
# 📂 DONNÉES RÉELLES — valeurs recopiées du fichier database.csv (les vraies commandes d'Amora Zone)
import plotly.express as px          # on ouvre l'appli de graphiques (une fois par session)

# Les 7 vraies commandes du 25 février 2025 (vues en M2.5) et leur heure de passage
heures_jour    = ["10:42", "12:32", "13:52", "14:57", "16:26", "21:04", "22:32"]
commandes_jour = [110.12, 116.05, 25.65, 46.18, 82.54, 32.45, 117.94]

print("x :", heures_jour)
print("y :", commandes_jour)
print("Même longueur ?", len(heures_jour) == len(commandes_jour))
````

### Un graphique en barres : `px.bar`
Une barre par commande : sa hauteur est le montant. 🖱️ Passe la souris sur une barre : Plotly affiche sa valeur exacte. Le graphique est **interactif**.

````python
# 📂 DONNÉES RÉELLES — valeurs recopiées du fichier database.csv (les vraies commandes d'Amora Zone)
plotly_commandes_jour = px.bar(
    x=heures_jour,                   # en bas : l'heure de chaque commande
    y=commandes_jour,                # la hauteur : son montant
    title="Les 7 commandes du 25 février 2025",
    labels={"x": "Heure de la commande", "y": "Montant (€)"},
)
plotly_commandes_jour.show()
````

👉 Lecture business : **4 barres dépassent 50 €** (la livraison offerte, calculée en M2.8), et les deux dernières commandes arrivent le soir, sur le site.

# [OPTIONNEL] M2.10+ · Le mini-CRM : boucler sur des fiches clients
**Le problème business :** appliquer la règle de statut de l'exercice 7 à **tous** les clients, puis compter les VIP et calculer le chiffre d'affaires total.

On combine tout : une **liste de dictionnaires** (M2.6), une **condition** (M2.7) et des **accumulateurs** (M2.8). C'est le cas le plus courant en entreprise : on parcourt les fiches clients une par une.

````python
# 🧪 EXEMPLE — données inventées pour comprendre, sans lien avec database.csv
clients = [
    {"prenom": "Léa",    "nb_commandes": 5,  "montant_total": 372.40},
    {"prenom": "Karim",  "nb_commandes": 1,  "montant_total": 45.90},
    {"prenom": "Inès",   "nb_commandes": 12, "montant_total": 1310.00},
    {"prenom": "Hugo",   "nb_commandes": 2,  "montant_total": 88.00},
    {"prenom": "Sarah",  "nb_commandes": 15, "montant_total": 1650.50},
    {"prenom": "Yanis",  "nb_commandes": 1,  "montant_total": 32.50},
]

for client in clients:
    print(client["prenom"], "→", client["montant_total"], "€")
````

👉 À chaque tour, `client` contient **une fiche entière** (un dictionnaire). On lit ses informations avec leur clé : `client["prenom"]`, `client["montant_total"]`.

### ✏️ Exercice 9 — Le mini-CRM
Pour chaque client de la liste `clients` :

- `nb_vip = nb_vip + 1` compte, comme `nb_livraisons_offertes` en M2.8. Elle est **dans** le `if` : on ne clique que pour les VIP.
- `ca_total = ca_total + client["montant_total"]` additionne : « le nouveau CA total, c'est l'ancien plus ce qu'a dépensé ce client ». Elle est **sous le `for` mais en dehors du `if`** : le chiffre d'affaires concerne **tous** les clients.

Le bon réflexe pour placer une ligne : **« ça concerne tous les clients, ou seulement certains ? »** La réponse donne son niveau de décalage.

💡 *Indice :* résultat attendu : 2 VIP (Inès et Sarah) et un CA total de 3499.3 €. Si tu obtiens 78.4 €, ta ligne `ca_total` est décalée d'un cran de trop (dans le `else`) ; si tu obtiens 32.5 €, tu as oublié `ca_total +`.
